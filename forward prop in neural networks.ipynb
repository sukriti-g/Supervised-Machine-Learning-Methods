```python
import numpy as np

def sigmoid(z):
    return 1/(1 + np.exp(-z))

#building the neural network assuming we have to weights and biases already
#w is the weight matrix
#b is the bias vector

def dense(a_in, w, b):
    units = w.shape[1]
    a_out = np.zeros((units,))
    for j in range(units):
        a_out[j] = sigmoid(np.dot(w[:,j], a_in) + b[j])
    return a_out

def sequential(a_in, w1, b1, w2, b2):
    #depends on how many layers are in the neural network
    #consider 2 layers
    a1 = dense(a_in, w1, b1)
    a2 = dense(a1, w2, b2)
    return a2
```
