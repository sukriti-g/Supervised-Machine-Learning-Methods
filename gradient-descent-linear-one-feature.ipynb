```python
import numpy as np
import matplotlib.pyplot as plt

#load dataset
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

#algorithm
def compute_cost(x, y, w, b):
    cost = 0
    for i in range(len(x)):
        f_wb = w * x[i] + b
        cost += (f_wb - y[i])**2
    return (1/(2*len(x))) * cost

def compute_gradient(x, y, w, b):
    dj_dw = 0
    dj_db = 0
    for i in range(len(x)):
        f_wb = w * x[i] + b
        dj_dw += (f_wb - y[i]) * x[i]
        dj_db += (f_wb - y[i])
    dj_dw = dj_dw / len(x)
    dj_db = dj_db / len(x)
    return dj_dw, dj_db

def gradient_descent(x, y, w_in, b_in, alpha, tolerance, cost_function, gradient_function):
    J = []
    weights = []
    w = w_in
    b = b_in
    w_c = w + .1

    while abs(w_c - w) > tolerance:
        dj_dw, dj_db = gradient_function(x,y,w,b)
        w_c = w
        w -= (alpha * dj_dw)
        b -= (alpha * dj_db)
        J.append(cost_function(x,y,w,b))
        weights.append([w,b])

    return w, b, J, weights

#plotting the cost function (for intuition)
b=0 #assuming
w = np.linspace(-100,500,100)
plt.plot(w, compute_cost(x_train, y_train, w, b), color = 'yellow')
plt.title("Cost Function")
plt.xlabel("w")
plt.ylabel("J(w,b)")
plt.grid()

#testing gradient descent
w_in = 0
b_in = 0
alpha = 0.01
tolerance = 0.000001
w_final, b_final, J_plot, weights_plot = gradient_descent(x_train, y_train, w_in, b_in, alpha, tolerance, compute_cost, compute_gradient)

print(w_final, b_final)

#predicting the price for some data points
print(f"1000 sqft house prediction {w_final*1.0 + b_final:0.1f} Thousand dollars")
print(f"1200 sqft house prediction {w_final*1.2 + b_final:0.1f} Thousand dollars")
print(f"2000 sqft house prediction {w_final*2.0 + b_final:0.1f} Thousand dollars")

```
